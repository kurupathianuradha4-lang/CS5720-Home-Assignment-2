NLP preprocessing

```python
import nltk
nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('stopwords')

import string
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer
from nltk.tokenize import word_tokenize

def preprocess(sentence):
    # Step 1: Tokenize into words and punctuation
    tokens = word_tokenize(sentence)

    # Step 2: Remove stopwords and punctuation
    stop_words = set(stopwords.words("english"))
    filtered = [t for t in tokens
                if t.lower() not in stop_words and t not in string.punctuation]

    # Step 3: Stem each remaining word (Porter stemmer)
    stemmer = PorterStemmer()
    stemmed = [stemmer.stem(t) for t in filtered]

    print("Original Tokens:", tokens)
    print("Tokens Without Stopwords:", filtered)
    print("Stemmed Words:", stemmed)

preprocess("NLP techniques are used in virtual assistants like Alexa and Siri.")
```

Scaled dot-product attention

```python
import numpy as np

def softmax(x):
    # Subtract the row max for numerical stability
    shifted = x - np.max(x, axis=-1, keepdims=True)
    exp = np.exp(shifted)
    return exp / np.sum(exp, axis=-1, keepdims=True)

def scaled_dot_product_attention(Q, K, V):
    d = K.shape[-1]               # key dimension
    scores = Q @ K.T              # 1) dot product of Q and K^T
    scaled = scores / np.sqrt(d)  # 2) scale by sqrt(d)
    weights = softmax(scaled)     # 3) softmax -> attention weights
    output = weights @ V          # 4) multiply weights by V
    return weights, output

Q = np.array([[1, 0, 1, 0], [0, 1, 0, 1]])
K = np.array([[1, 0, 1, 0], [0, 1, 0, 1]])
V = np.array([[1, 2, 3, 4], [5, 6, 7, 8]])

weights, output = scaled_dot_product_attention(Q, K, V)

np.set_printoptions(precision=4, suppress=True)
print("Attention weights matrix:\n", weights)
print("\nFinal output matrix:\n", output)
```
